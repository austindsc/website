## Polar Axis

For a demonstration of a line plot on a polar axis, see @fig-polarrrr

```python
#| label: fig-polarrrr
#| fig-cap: "a cool spiral"

import numpy as np
import matplotlib.pyplot as plt

r = np.arange(0, 2, 0.01)
theta = 2 * np.pi * r

fig, ax = plt.subplots(subplot_kw={'projection': 'polar'})
ax.plot(theta, r, color = 'red')
ax.set_rticks([0.5, 1, 1.5, 2])  # Less radial ticks


ax.grid(True)  # Turn on the grid for better visibility
plt.show()
```
